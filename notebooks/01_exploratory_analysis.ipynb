```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import requests
```

```python
import os
import requests
from dotenv import load_dotenv

# Load environment variables from .env
load_dotenv("../.env")

# Retrieve API key
api_key = os.getenv("RAPIDAPI_KEY")

# Verify the key loaded without revealing it
print("API key loaded:", bool(api_key))
```

```python
url = "https://jsearch.p.rapidapi.com/search-v2"

headers = {
    "X-RapidAPI-Key": api_key,
    "X-RapidAPI-Host": "jsearch.p.rapidapi.com"
}

params = {
    "query": "data science intern in Canada",
    "country": "ca",
    "num_pages": 1,
    "employment_types": "INTERN"
}

response = requests.get(
    url,
    headers=headers,
    params=params,
    timeout=30
)

print("Status code:", response.status_code)
```

```python
data = response.json()
jobs = data["data"]["jobs"]

print("Jobs retrieved:", len(jobs))

for job in jobs:
    print(job["job_title"], "-", job["employer_name"])
```

```python
import pandas as pd

df = pd.DataFrame(jobs)

print(df.shape)
df.head()
```
